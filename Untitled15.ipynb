```python
# 1. Required packages install karein
!pip install -q gradio transformers torch pillow sentencepiece tiktoken

import gradio as gr
from transformers import TrOCRProcessor, VisionEncoderDecoderModel, RobertaTokenizer, ViTImageProcessor
import torch

# Base model path
model_path = 'microsoft/trocr-base-printed'

print("Loading processor and model...")

# Direct RobertaTokenizer use kar rahe hain (is se sentencepiece error kabhi nahi aayega)
tokenizer = RobertaTokenizer.from_pretrained(model_path)
feature_extractor = ViTImageProcessor.from_pretrained(model_path)
processor = TrOCRProcessor(image_processor=feature_extractor, tokenizer=tokenizer)

# Model load karein
model = VisionEncoderDecoderModel.from_pretrained(model_path)
model.eval()

print("Model loaded successfully!")

def extract_urdu_text(image):
    if image is None:
        return 'Please upload an image'

    # Image process karein
    pixel_values = processor(image, return_tensors='pt').pixel_values

    with torch.no_grad():
        generated_ids = model.generate(pixel_values)

    text = processor.batch_decode(generated_ids, skip_special_tokens=True)[0]
    return text if text else 'Could not extract text from this image'

# Gradio Interface
interface = gr.Interface(
    fn=extract_urdu_text,
    inputs=gr.Image(type='pil', label='Upload Urdu Image'),
    outputs=gr.Textbox(label='Extracted Urdu Text'),
    title='Urdu OCR Code Saviours SI-26',
    description='Upload an image containing Urdu text and get the extracted text.'
)

# App launch karein
interface.launch(share=True)
```
